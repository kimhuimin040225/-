```python
import pandas as pd
from scipy import stats
from google.colab import files
uploaded = files.upload()
# 1. 데이터 불러오기 (정확한 경로로 수정 완료)
# 파일이 sample_data 폴더 안의 type3 폴더에 있으므로 아래와 같이 작성합니다.
df = pd.read_csv("before_after_worktime.csv")

# 2. 변수 설정
before = df["before_min"]
after = df["after_min"]

# 3. 대응표본 t-test 수행
# alternative="greater"는 교육 전(before) 시간이 교육 후(after)보다 더 길었는가(즉, 시간이 단축되었는가)를 검정합니다.
diff = before - after
result = stats.ttest_rel(before, after, alternative="greater")

# 4. 결과 출력
print("-" * 40)
print("   대응표본 t-test(Paired t-test) 결과   ")
print("-" * 40)
print(f"평균 차이(before - after): {diff.mean():.4f}분")
print(f"t-통계량(t-statistic): {result.statistic:.4f}")
print(f"p-값(p-value): {result.pvalue:.6f}")
print("-" * 40)

# 5. 결과 해석 가이드
if result.pvalue < 0.05:
    print("해석: p-value가 0.05보다 작으므로 귀무가설을 기각합니다.")
    print("결론: 교육 전후 업무처리시간에 유의미한 차이가 있으며, 교육 후 시간이 단축되었다고 볼 수 있습니다.")
else:
    print("해석: p-value가 0.05보다 크므로 귀무가설을 기각할 수 없습니다.")
    print("결론: 교육 전후 업무처리시간에 유의미한 차이가 있다고 보기 어렵습니다.")
```
